```python

customers = pd.DataFrame({
    "customer_id": [1, 2, 3],
    "name": ["Ali", "Sara", "Ahmed"],
    "plan": ["Pro", "Free", "Pro"]
})

orders = pd.DataFrame({
    "customer_id": [1, 1, 3],
    "order_id": ["O101", "O102", "O103"],
    "amount": [500, 300, 700]
})
```

```python
merged = customers.merge(orders, on="customer_id", how="inner")
print(merged)
```

```python
merged = customers.merge(orders, on="customer_id", how="left")
print(merged)
```

```python
merged = customers.merge(orders, on="customer_id", how="right")
print(merged)
```

```python
merged = customers.merge(orders, on="customer_id", how="outer")
print(merged)
```

```python
customers = pd.DataFrame({
    "customer_id": [1, 2, 3],
    "name": ["Ali", "Sara", "Ahmed"]
})

orders = pd.DataFrame({
    "user_id": [1, 1, 3],
    "amount": [500, 300, 700]
})
```

```python
merged = customers.merge(orders, left_on="customer_id", right_on="user_id", how="inner")
print(merged)
```

```python
customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Ali", "Sara", "Ahmed", "Hina"],
    "plan": ["Pro", "Free", "Pro", "Free"]
})

orders = pd.DataFrame({
    "customer_id": [1, 1, 3, 5],
    "order_id": ["O101", "O102", "O103", "O104"],
    "amount": [500, 300, 700, 200]
})
```

```python
merged_df = customers.merge(orders, on="customer_id", how="left")
print(merged_df)
```

```python
nan_rows = merged_df[merged_df['order_id'].isna()]
print(nan_rows)
```

```python
merged_df = customers.merge(orders, on="customer_id", how="right")
print(merged_df)
```

```python
nan_rows = merged_df[merged_df['customer_id'].isna()]
print(nan_rows)
```

```python
# 1. Right join keeps everything from the orders table
right_merged = customers.merge(orders, on="customer_id", how="right")

# 2. Filter where customers table columns (like customer_name or email) are NaN
orders_only = right_merged[right_merged['customer_name'].isna()]

print(orders_only['customer_id'].unique())
```
