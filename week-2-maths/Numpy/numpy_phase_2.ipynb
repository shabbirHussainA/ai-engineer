```python
arr_2d = np.array([[1,2,3],[4,5,6]])
print(arr_2d[1,2]) #2nd row 3rd digit
print(arr_2d[1]) # complete second row
print(arr_2d[:,2]) # all rows 3rd columne
```

```python
arr_2d_unsorted = np.array([[1,3,2],[4,1,7]])
print(arr_2d_unsorted)
arr_2d_sorted = np.sort(arr_2d_unsorted,1)
print(arr_2d_sorted)
```

```python
numbers = np.arange(1,13)
even_numbers = numbers[numbers %2 == 0]
print(even_numbers)
```

```python
mask = numbers > 5
print(mask)

print(numbers[mask])
```

```python
indices = [0,2,4]
print(numbers[indices])
```

```python
where_result = np.where(numbers > 5)
print(where_result)
print(numbers[where_result])
```

```python
condition_array = np.where(numbers > 5, numbers*2, numbers)
print(condition_array)
```

```python
original = np.array([[1,2,3],[4,5,6]])
new_row = np.array([10,20,30])
with_new_row = np.vstack((original,new_row))
print(original)
print(with_new_row)
```

```python
new_col = np.array([[100,200],[300,400]])
with_new_col = np.column_stack((original,new_col))
print(with_new_col)
```
